### Exercício Proposto 1
### Exercício Proposto 1: Percentual de Ocupação de Garagem
Com base no exemplo do Plano Diretor de Desenvolvimento Urbano apresentado na aula, escreva um programa em Python que receba as dimensões (largura e profundidade) de uma garagem e de um terreno. O programa deve calcular a área de cada um, calcular o percentual de ocupação da garagem em relação ao terreno e exibir o resultado formatado.

**Requisitos:**
- Solicitar entradas do usuário usando `input()` convertidos para `float`.
- Calcular as áreas multiplicando largura por profundidade.
- Calcular a porcentagem: `(Area_Garagem / Area_Terreno) * 100`.
- Exibir o resultado final.


```python
largura_garagem = float(input("Largura da garagem: "))
profundidade_garagem= float(input("Profundidade da garagem: "))

largura_terreno = float(input("Largura da terreno: "))
profundidade_terreno= float(input("Profundidade da terreno: "))

Area_garagem= largura_garagem * profundidade_garagem
Area_terreno= largura_terreno * profundidade_terreno

percentual_ocupacao = (Area_garagem / Area_terreno) * 100

print(f"A área da garagem e do terreno são, respectivamente: {Area_garagem:.2f} m², {Area_terreno:.2f} m²")
print(f"O percentual de ocupação da garagem em relação ao terreno é {percentual_ocupacao:.2f}%")
```

### Exercício Proposto 2
### Exercício Proposto 2: Cálculo de Média e Aprovação de Alunos
Implemente o algoritmo de cálculo de médias apresentado na página 23 da apostila. O programa deve receber o nome de um aluno, suas notas de duas provas (P1 e P2), calcular a média aritmética simples e determinar se o aluno está "Aprovado" (média maior ou igual a 7.0) ou em "Exame" (média menor que 7.0).

**Requisitos:**
- Receber o nome do aluno (string).
- Receber as notas P1 e P2 (float).
- Calcular a média.
- Utilizar a estrutura condicional `if/else` para classificar o aluno.


```python
# Comece seu código aqui
nome = input("Nome do aluno: ")
nota_p1 = float(input("Nota P1: "))
nota_p2 = float(input("Nota P2: "))

media = (nota_p1 + nota_p2) / 2

if media >= 7.0:
  print(f"Aluno {nome}: Aprovado")
else:
  print(f"Aluno {nome}: Exame")
```

**Requisitos:**
- Solicitar o custo de fabricação (float).
- Solicitar a margem de lucro em porcentagem (float).

```python
# Entrada de dados
custo_base = float(input("Digite o custo de produção do calçado (R$): "))
percentual_imposto = float(input("Digite a alíquota de impostos (%): "))
percentual_lucro = float(input("Digite a margem de lucro desejada (%): "))

# Cálculos intermediários
valor_imposto = custo_base * (percentual_imposto / 100)
custo_com_imposto = custo_base + valor_imposto
valor_lucro = custo_com_imposto * (percentual_lucro / 100)

# Preço final de venda
preco_venda = custo_com_imposto + valor_lucro

# Exibição do relatório detalhado
print("\n=========================================")
print("       RELATÓRIO DE VIABILIDADE          ")
print("=========================================")
print(f"Custo de Produção Base: R$ {custo_base:.2f}")
print(f"Impostos ({percentual_imposto:.1f}%): R$ {valor_imposto:.2f}")
print(f"Custo Acumulado: R$ {custo_com_imposto:.2f}")
print(f"Lucro Esperado ({percentual_lucro:.1f}%): R$ {valor_lucro:.2f}")
print(f"Preço Final de Venda: R$ {preco_venda:.2f}")
print("-----------------------------------------")

# Avaliação de competitividade
limite_competitivo = 150.00
if preco_venda <= limite_competitivo:
    print("STATUS: PRODUTO ALTAMENTE COMPETITIVO NO MERCADO!")
else:
    print(f"STATUS: FORA DA MARGEM COMPETITIVA (Excede R$ {limite_competitivo:.2f}). Sugere-se revisar as margens.")
print("=========================================")
```
